```python
tree = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H', 'I'],
    'E': ['J', 'K'],
    'F': ['L', 'M'],
    'G': ['N', 'O'],
}

payoffs = {'H': 3, 'I': 12, 'J': 8, 'K': 2, 'L': 4, 'M': 6, 'N': 14, 'O': 5}


def minimax(node, is_max, counter, reverse=False):
  counter[0] += 1
  if node in payoffs:
    return payoffs[node]

  children = list(reversed(tree[node])) if reverse else tree[node]

  if is_max:
    return max(minimax(child, False, counter, reverse) for child in children)
  else:
    return min(minimax(child, True, counter, reverse) for child in children)


def alphabeta(node, alpha, beta, is_max, counter, reverse=False):
  counter[0] += 1
  if node in payoffs:
    return payoffs[node]

  children = list(reversed(tree[node])) if reverse else tree[node]

  if is_max:
    val = float('-inf')
    for child in children:
      val = max(val, alphabeta(child, alpha, beta, False, counter, reverse))
      alpha = max(alpha, val)
      if alpha >= beta:
        break
    return val
  else:
    val = float('inf')
    for child in children:
      val = min(val, alphabeta(child, alpha, beta, True, counter, reverse))
      beta = min(beta, val)
      if alpha >= beta:
        break
    return val


c_mm1, c_ab1 = [0], [0]
val_mm1 = minimax('A', True, c_mm1, reverse=False)
val_ab1 = alphabeta(
    'A', float('-inf'), float('inf'), True, c_ab1, reverse=False
)

print('=== СОЛДАН ОҢҒА ҚАРАЙ ===')
print(f'Minimax мәні: {val_mm1}, Есептелген түйін саны: {c_mm1[0]}')
print(f'Alpha-Beta мәні: {val_ab1}, Есептелген түйін саны: {c_ab1[0]}')

c_mm2, c_ab2 = [0], [0]
val_mm2 = minimax('A', True, c_mm2, reverse=True)
val_ab2 = alphabeta('A', float('-inf'), float('inf'), True, c_ab2, reverse=True)

print('\n=== ОҢНАН СОЛҒА ҚАРАЙ ===')
print(f'Minimax мәні: {val_mm2}, Есептелген түйін саны: {c_mm2[0]}')
print(f'Alpha-Beta мәні: {val_ab2}, Есептелген түйін саны: {c_ab2[0]}')
```
