1-тапсырма

```python
def hill_climbing(values, start):
  current = start
  while True:
     neighbors = []
     if current > 0:
         neighbors.append(current - 1)
     if current < len(values) - 1:
         neighbors.append(current + 1)
     best = current
     for n in neighbors:
         if values[n] > values[best]:
             best = n
     if best == current:
         break
     current = best
  return current
values = [1, 3, 5, 8, 6, 4, 2]
result = hill_climbing(values, 4)
print("Соңғы индекс:", result)
print("Соңғы мəн:", values[result])
```

|---|---|---|---|---|
| 1 | 4 | 6 | 8 және 4 |  8 |
| 2 | 3 | 8 | 5 пен 6 | Тоқтайды |

4-тапсырма

```python
import random

def random_restart(values, attempts):
    best_position = None
    for _ in range(attempts):
        start = random.randint(0, len(values) - 1)
        position = hill_climbing(values, start)
        if best_position is None or values[position] > values[best_position]:
            best_position = position
    return best_position
values = [1, 4, 7, 5, 3, 6, 9, 8]
for experiment in range(1, 4):
    result = random_restart(values, 5)
    print("Эксперимент", experiment)
    print("Ең жақсы индекс:", result)
    print("Ең жақсы мән:", values[result])
    print("-" * 25)
```

Random Restart іздеуді әртүрлі кездейсоқ нүктелерден бастайды. Бір нүкте жергілікті максимумда (7-де) тіреліп қалса да, басқа нүкте глобал максимумға (9-ға) жеткізеді. Сондықтан ол жалғыз бастапқы орынға тәуелді болмайды.

5-тапсырма

```python
import math
```

```python
def random_neighbor(current, n):
    neighbors = []
    if current > 0:
        neighbors.append(current - 1)
    if current < n - 1:
        neighbors.append(current + 1)
    return random.choice(neighbors)
```

```python
def anneal(values, start, T=10):
    current = start
    while T > 0.1:
        next_state = random_neighbor(current, len(values))
        dE = values[next_state] - values[current]
        if dE > 0 or random.random() < math.exp(dE / T):
            current = next_state
        T *= 0.95
    return current
```

```python
values = [1, 4, 7, 5, 3, 6, 9, 8]
results = []
for experiment in range(1, 6):
    result = anneal(values, 0)
    results.append(values[result])
    print("Іске қосу", experiment)
    print("Соңғы индекс:", result)
    print("Соңғы мән:", values[result])
    print()
print("Барлық соңғы мәндер:", results)

if len(set(results)) == 1:
    print("Бес нәтиженің барлығы бірдей.")
else:
    print("Бес нәтиженің барлығы бірдей емес.")
```

Бес нәтиженің барлығы бірдей емес. Simulated Annealing алгоритмінде көрші күй кездейсоқ таңдалады және ол жақсы да, нашар да қадам болуы мүмкін. Сондықтан әр іске қосқан сайын алгоритм әртүрлі жолмен жүріп, әртүрлі нәтиже береді.
