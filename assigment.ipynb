```python
states = ["A", "B", "C", "D", "E", "S", "F", "H", "I"]
actions = ["UP", "DOWN", "LEFT", "RIGHT"]

gamma = 0.9
theta = 0.001

positions = {
    "A": (0, 0), "B": (0, 1), "C": (0, 2), "G": (0, 3),
    "D": (1, 0),              "E": (1, 2), "X": (1, 3),
    "S": (2, 0), "F": (2, 1), "H": (2, 2), "I": (2, 3)
}
state_at = {pos: state for state, pos in positions.items()}

moves = {
    "UP": (-1, 0), "DOWN": (1, 0),
    "LEFT": (0, -1), "RIGHT": (0, 1)
}

outcomes = {
    "UP":    [("UP", 0.8), ("LEFT", 0.1), ("RIGHT", 0.1)],
    "DOWN":  [("DOWN", 0.8), ("RIGHT", 0.1), ("LEFT", 0.1)],
    "LEFT":  [("LEFT", 0.8), ("DOWN", 0.1), ("UP", 0.1)],
    "RIGHT": [("RIGHT", 0.8), ("UP", 0.1), ("DOWN", 0.1)]
}

def move(state, direction):
    row, col = positions[state]
    dr, dc = moves[direction]
    new_pos = (row + dr, col + dc)
    if new_pos not in state_at:
        return state
    return state_at[new_pos]

def reward(next_state):
    if next_state == "G":
        return 10.0
    if next_state == "X":
        return -10.0
    return -0.1

V = {s: 0.0 for s in states}

while True:
    delta = 0.0
    new_V = {}
    for s in states:
        q_values = []
        for a in actions:
            q = 0.0
            for direction, prob in outcomes[a]:
                next_s = move(s, direction)
                r = reward(next_s)
                next_v = 0.0 if next_s in ["G", "X"] else V[next_s]
                q += prob * (r + gamma * next_v)
            q_values.append(q)
        new_V[s] = max(q_values)
        delta = max(delta, abs(new_V[s] - V[s]))
    V = new_V
    if delta < theta:
        break

policy = {}
for s in states:
    best_a = None
    best_q = float("-inf")
    for a in actions:
        q = 0.0
        for direction, prob in outcomes[a]:
            next_s = move(s, direction)
            r = reward(next_s)
            next_v = 0.0 if next_s in ["G", "X"] else V[next_s]
            q += prob * (r + gamma * next_v)
        if q > best_q:
            best_q = q
            best_a = a
    policy[s] = best_a

print("FINAL STATE VALUES:")
for s in sorted(V.keys()):
    print(f"{s}: {V[s]:.3f}")

print("\nFINAL POLICY:")
for s in sorted(policy.keys()):
    print(f"{s} -> {policy[s]}")
```
