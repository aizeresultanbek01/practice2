```python
graph = {
    'A': [('B', 2), ('C', 5), ('E', 3)],
    'B': [('D', 4), ('E', 3)],
    'C': [('E', 4)],
    'D': [('E', 2), ('G', 5)],
    'E': [('G', 3)],
    'G': []
}
heuristic = {'A': 6, 'B': 4, 'C': 5, 'D': 4, 'E': 2, 'G': 0}

def dfs(graph, start, goal):
    stack = [(start, [start])]
    visited = []

    while len(stack) > 0:
        node, path = stack.pop()
        if node not in visited:
            visited.append(node)
            if node == goal:
                return path, visited
            for neighbor, cost in reversed(graph[node]):
                stack.append((neighbor, path + [neighbor]))

def bfs(graph, start, goal):
    queue = [(start, [start])]
    visited = []

    while len(queue) > 0:
        node, path = queue.pop(0)

        if node not in visited:
            visited.append(node)

            if node == goal:
                return path, visited
            for neighbor, cost in graph[node]:
                queue.append((neighbor, path + [neighbor]))

def ucs(graph, start, goal):
    queue = [(0, start, [start])]
    visited = []

    while len(queue) > 0:
        queue.sort()
        g, node, path = queue.pop(0)

        if node not in visited:
            visited.append(node)

            if node == goal:
                return path, g, visited

            for neighbor, cost in graph[node]:
                queue.append((g + cost, neighbor, path + [neighbor]))


def astar(graph, start, goal, h):
    queue = [(h[start], 0, start, [start])]
    visited = []

    while len(queue) > 0:
        queue.sort()
        f, g, node, path = queue.pop(0)
        if node not in visited:
            visited.append(node)
            if node == goal:
                return path, g, visited
            for neighbor, cost in graph[node]:
                new_g = g + cost
                new_f = new_g + h[neighbor]  # f = g + h
                queue.append((new_f, new_g, neighbor, path + [neighbor]))

print("DFS жолы:", dfs(graph, 'A', 'G')[0])
print("BFS жолы:", bfs(graph, 'A', 'G')[0])

ucs_path, ucs_cost, _ = ucs(graph, 'A', 'G')
print(f"UCS жолы: {ucs_path} (Құны: {ucs_cost})")

astar_path, astar_cost, _ = astar(graph, 'A', 'G', heuristic)
print(f"A* жолы:  {astar_path} (Құны: {astar_cost})")
```
