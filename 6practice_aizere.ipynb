```python
def bayes(prior, p_obs_if_obj, p_obs_if_clr):
    score_obj = prior * p_obs_if_obj
    score_clr = (1.0 - prior) * p_obs_if_clr
    return score_obj / (score_obj + score_clr)

# 1. A бөлімі: Бір реттік есептеулер
print("A) BLOCKED:", round(bayes(0.30, 0.80, 0.20), 4))
print("A) CLEAR:  ", round(bayes(0.30, 0.20, 0.80), 4))

# 2. D1 эксперименті: Бастапқы ықтималдықты өзгерту
print("\n--- D1: Бастапқы ықтималдықтар ---")
for p in [0.10, 0.30, 0.50, 0.70]:
    print(f"Prior {p} -> {bayes(p, 0.80, 0.20):.4f}")

# 3. D2 эксперименті: Жалған дабылды арттыру (0.20 -> 0.40)
print("\n--- D2: Жалған дабыл (0.40) ---")
print(f"Nәтиже: {bayes(0.30, 0.80, 0.40):.4f}")

# 4. HMM функциясы (C және D3 үшін ортақ)
def run_hmm(obs_list):
    belief = 0.30
    print("t  reading  pred_O  upd_O   upd_C")
    for t, r in enumerate(obs_list, 1):
        pred = belief if t == 1 else 0.70 * belief + 0.20 * (1.0 - belief)
        like_o, like_c = (0.80, 0.20) if r == "BLOCKED" else (0.20, 0.80)
        belief = bayes(pred, like_o, like_c)
        print(f"{t}  {r:7}  {pred:.4f}  {belief:.4f}  {1.0-belief:.4f}")

# 5. C бөлімі (Негізгі HMM тізбегі)
print("\n--- C бөлімі: HMM негізгі тізбегі ---")
run_hmm(["BLOCKED", "CLEAR", "BLOCKED", "BLOCKED"])

# 6. D3 эксперименті (Басқа тізбек)
print("\n--- D3 Эксперименті: Басқа бақылаулар ---")
run_hmm(["CLEAR", "CLEAR", "BLOCKED", "CLEAR"])
```
